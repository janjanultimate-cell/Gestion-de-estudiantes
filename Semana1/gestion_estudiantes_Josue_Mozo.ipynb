```python

#FUNCION PARA AGREGAR ESTUDIANTES
estudiantes = []

def almacenar_estudiantes(nombre, edad, calificaciones):
    estudiante = {
        "nombre":nombre,
        "edad":edad,
        "calificaciones":calificaciones
    }

    estudiantes.append(estudiante)

almacenar_estudiantes("Josue",18,[7,6,3]),
almacenar_estudiantes("Ezequiel",19,[10,9,8]),
almacenar_estudiantes("Mozo",20,[7,7,8])


# FUNCION PARA CALCULAR EL PROMEDIO DE LOS ESTUDIANTES
def promedio(estudiante):
    calificaciones=estudiante["calificaciones"]
    promedio_estudiante = sum(calificaciones) / len(calificaciones)
    return promedio_estudiante

print("NOMBRE  | PROMEDIO")

for estudiante in estudiantes:
    resultado = promedio(estudiante)
    
    print(estudiante["nombre"],":",round(resultado,2))

print("--EL ESTUDIANTE CON MEJOR PROMEDIO ES:--")

# FUNCION PARA CALCULAR EL MEJOR PROMEDIO DE LOS ESTUDIANTES

def mejor_promedio():
    mejor_promedio_estudiante=estudiantes[0]

    for estudiante in estudiantes:
        if promedio(estudiante)>promedio(mejor_promedio_estudiante):
             mejor_promedio_estudiante = estudiante
    return mejor_promedio_estudiante

mejor_promedio_estudiante = mejor_promedio()

print(mejor_promedio_estudiante["nombre"])

# FUNCION PARA CALCULAR EL PEOR PROMEDIO DE LOS ESTUDIANTES

print("--EL ESTUDIANTE CON PEOR PROMEDIO ES:--")

def peor_promedio():
    peor_promedio_estudiante=estudiantes[0]

    for estudiante in estudiantes:
        if promedio(estudiante)<promedio(peor_promedio_estudiante):
             peor_promedio_estudiante = estudiante
    return peor_promedio_estudiante

peor_promedio_estudiante = peor_promedio()

print(peor_promedio_estudiante["nombre"])


# FUNCION PARA CALCULAR EL PROMEDIO GENERAL

def promedio_general():
    suma_de_promedios = 0

    for estudiante in estudiantes:
        suma_de_promedios+=promedio(estudiante)
    return suma_de_promedios / len(estudiantes)

resultado_promedio = promedio_general()


print("--EL PROMEDIO GENERAL:--")

print("el promedio general: ", round(resultado_promedio,2))
```
